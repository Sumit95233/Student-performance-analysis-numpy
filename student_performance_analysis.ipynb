### Student_Performance_Analysis

```python
# STEP 1: IMPORT LIBRARIES
import numpy as np
import csv

# STEP 2: LOAD CSV WITH ENCODING SUPPORT
file_path = "student_analysis_data.csv"

# Try different encodings
encodings = ["utf-8-sig", "utf-16", "cp1252", "latin-1"]

data = None

for encoding in encodings:

    try:
        with open(
            file_path,
            "r",
            encoding=encoding,
            newline=""
        ) as file:

            reader = csv.DictReader(file)

            temp_data = list(reader)

            if temp_data and reader.fieldnames:
                data = temp_data
                print("File Encoding:", encoding)
                break

    except UnicodeError:
        continue

# Check whether dataset loaded
if data is None:
    raise ValueError("Unable to read CSV file. Check file format.")

# Clean column headers
data_headers = [
    header.strip().replace("\ufeff", "")
    for header in reader.fieldnames
]

# STEP 3: VERIFY DATASET

print("\nDATASET LOADED SUCCESSFULLY!")

print("Total Records:", len(data))

print("Total Columns:", len(data_headers))

print("Column Names:", data_headers)
```

```python

# STEP 2: CONVERT DATA INTO NUMPY ARRAYS

student_id = np.array([row["student_id"] for row in data])
names = np.array([row["name"] for row in data])
emails = np.array([row["email"] for row in data])

ages = np.array([int(row["age"]) for row in data])
genders = np.array([row["gender"] for row in data])
departments = np.array([row["department"] for row in data])
semesters = np.array([int(row["semester"]) for row in data])
cities = np.array([row["city"] for row in data])

# Extract subject marks

maths = np.array([int(row["mathematics"]) for row in data])
physics = np.array([int(row["physics"]) for row in data])
programming = np.array([int(row["programming"]) for row in data])
dbms = np.array([int(row["database_management"]) for row in data])
english = np.array([int(row["english"]) for row in data])

# Combine all subject marks into one NumPy array

subjects = np.array([
    "Mathematics",
    "Physics",
    "Programming",
    "DBMS",
    "English"
])

marks = np.column_stack((
    maths, physics, programming, dbms, english
))

# Calculate total marks and percentage

total_marks = np.sum(marks, axis=1)

percentage = (total_marks / 500) * 100

# Display array information

print("NUMPY ARRAYS CREATED SUCCESSFULLY!")

print("Student Array Shape:", student_id.shape)

print("Marks Array Shape:", marks.shape)

print("Percentage Array Shape:", percentage.shape)

print("\nFirst 5 Student Records:")

for i in range(5):
    print(names[i], total_marks[i], round(percentage[i], 2))
```

```python
# Q1: STUDENT DEMOGRAPHIC ANALYSIS

# Calculate total number of students
total_students = np.size(student_id)

# Calculate average age
average_age = np.mean(ages)

# Find minimum and maximum age
minimum_age = np.min(ages)
maximum_age = np.max(ages)

# Calculate gender distribution
unique_genders, gender_counts = np.unique(
    genders,
    return_counts=True
)

# Display results
print("STUDENT DEMOGRAPHIC ANALYSIS")

print("Total Students:", total_students)

print("Average Age:", round(average_age, 2))

print("Minimum Age:", minimum_age)

print("Maximum Age:", maximum_age)

print("\nGENDER DISTRIBUTION")

for gender, count in zip(unique_genders, gender_counts):
    print(gender, ":", count)
```

```python
# Q2: SUBJECT-WISE AVERAGE MARKS ANALYSIS

# Calculate average marks of each subject
average_marks = np.mean(marks, axis=0)

# Find subject with highest average marks
highest_subject_index = np.argmax(average_marks)

# Find subject with lowest average marks
lowest_subject_index = np.argmin(average_marks)

# Display subject-wise average marks
print("SUBJECT-WISE AVERAGE MARKS")

for i in range(len(subjects)):

    print(
        subjects[i],
        ":",
        round(average_marks[i], 2)
    )

# Display strongest subject


print("\nSTRONGEST SUBJECT:",subjects[highest_subject_index])

print("Average Marks:",
      round(average_marks[highest_subject_index], 2))

# Display weakest subject

print("\nWEAKEST SUBJECT:", subjects[lowest_subject_index])

print("Average Marks:",round(average_marks[lowest_subject_index], 2))
```

```python
# Q3: TOP 10 PERFORMING STUDENTS

# Sort students in descending order of total marks
top_indices = np.argsort(total_marks)[::-1][:10]

# Display top 10 students
print("TOP 10 PERFORMING STUDENTS")

for rank, index in enumerate(top_indices, start=1):

    print(
        "Rank:", rank,
        "| Student ID:", student_id[index],
        "| Name:", names[index],
        "| Total Marks:", total_marks[index],
        "| Percentage:", round(percentage[index], 2), "%"
    )
```

```python
# Q4: BOTTOM 10 PERFORMING STUDENTS

# Sort students in ascending order of total marks
bottom_indices = np.argsort(total_marks)[:10]

# Display bottom 10 students
print("BOTTOM 10 PERFORMING STUDENTS")

for rank, index in enumerate(bottom_indices, start=1):

    print(
        "Rank:", rank,
        "| Student ID:", student_id[index],
        "| Name:", names[index],
        "| Total Marks:", total_marks[index],
        "| Percentage:", round(percentage[index], 2), "%"
    )
```

```python
# Q5: CLASS TOPPER ANALYSIS

# Find index of student with highest total marks
topper_index = np.argmax(total_marks)

# Find highest total marks
highest_total = np.max(total_marks)

# Calculate topper percentage
topper_percentage = percentage[topper_index]

# Assign grade according to percentage
topper_grade = np.where(
    topper_percentage >= 90, "A",
    np.where(
        topper_percentage >= 80, "B",
        np.where(
            topper_percentage >= 70, "C",
            np.where(
                topper_percentage >= 60, "D",
                np.where(topper_percentage >= 50, "E", "F")
            )
        )
    )
)

# Display topper information
print("CLASS TOPPER ANALYSIS")

print("Student ID:", student_id[topper_index])

print("Name:", names[topper_index])

print("Email:", emails[topper_index])

print("Department:", departments[topper_index])

print("Age:", ages[topper_index])

print("Total Marks:", highest_total)

print("Percentage:", round(topper_percentage, 2), "%")

print("Grade:", topper_grade)

print("\nCLASS TOPPER IDENTIFIED SUCCESSFULLY!")
```

```python
# Q6: SUBJECT-WISE HIGHEST AND LOWEST MARKS

# Calculate highest marks in each subject
highest_marks = np.max(marks, axis=0)

# Calculate lowest marks in each subject
lowest_marks = np.min(marks, axis=0)

# Display subject-wise results
print("SUBJECT-WISE HIGHEST AND LOWEST MARKS")

for i in range(len(subjects)):

    # Find student with highest marks
    highest_index = np.argmax(marks[:, i])

    # Find student with lowest marks
    lowest_index = np.argmin(marks[:, i])

    print("\nSUBJECT:", subjects[i])

    print(
        "Highest Marks:", highest_marks[i],
        "| Student:", names[highest_index]
    )

    print(
        "Lowest Marks:", lowest_marks[i],
        "| Student:", names[lowest_index]
    )
```

```python
# Q7: STUDENT PASS/FAIL ANALYSIS

# Set minimum passing percentage
passing_percentage = 50

# Classify students based on overall percentage
results = np.where(
    percentage >= passing_percentage,
    "Pass",
    "Fail"
)

# Calculate total passed students
pass_count = np.sum(results == "Pass")

# Calculate total failed students
fail_count = np.sum(results == "Fail")

# Calculate pass and fail rates
pass_rate = np.mean(results == "Pass") * 100

fail_rate = np.mean(results == "Fail") * 100

# Display results
print("STUDENT PASS/FAIL ANALYSIS\n")

print("Total Students:", len(student_id))

print("Passing Criteria:", passing_percentage, "%")

print("Total Passed:", pass_count)

print("Total Failed:", fail_count)

print("Pass Rate:", round(pass_rate, 2), "%")

print("Fail Rate:", round(fail_rate, 2), "%")

# Verify total students
print("\nVerification:", pass_count + fail_count)
```

```python
# Q8: DEPARTMENT-WISE PERFORMANCE ANALYSIS

# Find all unique departments
unique_departments = np.unique(departments)

# Store department averages
department_averages = []

# Display department-wise performance
print("DEPARTMENT-WISE PERFORMANCE ANALYSIS")

for department in unique_departments:

    # Filter students belonging to department
    department_filter = departments == department

    # Calculate total students
    total_department_students = np.sum(department_filter)

    # Calculate average percentage
    department_average = np.mean(
        percentage[department_filter]
    )

    department_averages.append(department_average)

    print("\nDepartment:", department)

    print("Total Students:", total_department_students)

    print("Average Percentage:",
          round(department_average, 2), "%")

# Convert averages into NumPy array
department_averages = np.array(department_averages)

# Identify best-performing department
best_department_index = np.argmax(department_averages)

print("\nBEST PERFORMING DEPARTMENT")

print("Department:",
      unique_departments[best_department_index])

print("Average Percentage:",
      round(department_averages[best_department_index], 2), "%")
```

```python
# Q9: STATISTICAL ANALYSIS

# Calculate average percentage
mean_percentage = np.mean(percentage)

# Calculate median percentage
median_percentage = np.median(percentage)

# Calculate variance
variance = np.var(percentage)

# Calculate standard deviation
standard_deviation = np.std(percentage)

# Calculate minimum and maximum percentage
minimum_percentage = np.min(percentage)

maximum_percentage = np.max(percentage)

# Display statistical results
print("STUDENT PERFORMANCE STATISTICS\n")

print("Mean Percentage:",
      round(mean_percentage, 2), "%")

print("Median Percentage:",
      round(median_percentage, 2), "%")

print("Variance:",
      round(variance, 2))

print("Standard Deviation:",
      round(standard_deviation, 2))

print("Minimum Percentage:",
      round(minimum_percentage, 2), "%")

print("Maximum Percentage:",
      round(maximum_percentage, 2), "%")
```

```python
# Q10: GRADE DISTRIBUTION AND CLASSIFICATION

# Assign grades based on percentage
grades = np.where(
    percentage >= 90, "A",
    np.where(
        percentage >= 80, "B",
        np.where(
            percentage >= 70, "C",
            np.where(
                percentage >= 60, "D",
                np.where(
                    percentage >= 50, "E", "F"
                )
            )
        )
    )
)

# Classify students as Pass or Fail
results = np.where(
    percentage >= 50,
    "Pass",
    "Fail"
)

# Calculate grade distribution
unique_grades, grade_counts = np.unique(
    grades,
    return_counts=True
)

# Display grade distribution
print("STUDENT GRADE DISTRIBUTION")

for grade, count in zip(unique_grades, grade_counts):

    print("Grade", grade, ":", count, "Students")

# Calculate total passed and failed students
pass_count = np.sum(results == "Pass")

fail_count = np.sum(results == "Fail")

# Display final results
print("\nFINAL CLASSIFICATION")

print("Total Students:", len(student_id))

print("Total Passed:", pass_count)

print("Total Failed:", fail_count)

print("\nGRADE ANALYSIS COMPLETED SUCCESSFULLY!")
```
